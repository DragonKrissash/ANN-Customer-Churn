```python
import tensorflow as tf
import pandas as pd
import numpy as np
import pickle
from tensorflow.keras.models import load_model
```

```python
### Load the trained model
model=load_model('model.h5')
```

```python
with open('onehotencoder_geo.pkl','rb') as file:
    label_encoder_geo=pickle.load(file)
```

```python
with open('label_encoder_gender.pkl','rb') as file:
    label_encoder_gender=pickle.load(file)
with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
input_data={
    'CreditScore':600,
    'Geography':'France',
    'Gender':'Male',
    'Age':40,
    'Tenure':3,
    'Balance':60000,
    'NumOfProducts':2,
    'HasCrCard':1,
    'IsActiveMember':1,
    'EstimatedSalary':50000
}
```

```python
list(label_encoder_geo.get_feature_names_out())
```

```python
test_point=pd.DataFrame([input_data])
```

```python
label_encoder_geo.transform(test_point[['Geography']]).toarray()
```

```python
temp=pd.DataFrame(label_encoder_geo.transform(test_point[['Geography']]).toarray(),columns=list(label_encoder_geo.get_feature_names_out()))
```

```python
test_point=pd.concat([test_point.drop('Geography',axis=1),temp],axis=1)
```
